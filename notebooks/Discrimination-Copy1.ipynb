```python
file = uproot.open("/pbs/throng/training/nantes-m2-rps-exp/data/run291263.data.root")
file.keys()
```

```python
events = file["eventsTree"]

m = events.arrays(
    ["nMuons", "Muon_Px", "Muon_Py", "Muon_Pz", "Muon_Charge", "Muon_E","isCMUL","zVtx","Muon_matchedTrgThreshold","isCMSH","isCMSL"],
    how="zip",
)
print(len(m))
print(m[3].to_list())
```

```python
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt

# Filtrer les événements  & (m.isCMUL)
filtered_events = m[(m.nMuons >= 2) & (abs(m.zVtx) < 10)]
print("Nombre de particules considérées :", len(filtered_events))

# Initialiser les vecteurs des muons
muon_vectors = ak.zip({
    "px": filtered_events.Muon.Px,
    "py": filtered_events.Muon.Py,
    "pz": filtered_events.Muon.Pz,
    "E": filtered_events.Muon.E,
    "charge": filtered_events.Muon.Charge
}, with_name="Momentum4D")

# Combinaisons de paires
muon_pairs = ak.combinations(muon_vectors, 2, fields=["muon1", "muon2"])

# Paires de charges opposées et égales
opposite_charge_pairs = muon_pairs[muon_pairs.muon1.charge != muon_pairs.muon2.charge]
same_charge_pairs = muon_pairs[muon_pairs.muon1.charge == muon_pairs.muon2.charge]
# Calcul des masses invariantes
masses_opposite = (opposite_charge_pairs.muon1 + opposite_charge_pairs.muon2).mass
masses_same = (same_charge_pairs.muon1 + same_charge_pairs.muon2).mass

# Aplatir les données et filtrer
flattened_opposite = ak.flatten(masses_opposite[(masses_opposite >= 1.5) & (masses_opposite <= 10)], axis=None)
flattened_same = ak.flatten(masses_same[(masses_same >= 1.5) & (masses_same <= 10)], axis=None)
print(len(flattened_opposite),len(flattened_same), "rapport des deux", len(flattened_opposite)/len(flattened_same))

# Calcul des histogrammes avec numpy
bins = 100
hist_opposite, bin_edges = np.histogram(flattened_opposite, bins=bins, range=(1.5, 5))
hist_same, _ = np.histogram(flattened_same, bins=bins, range=(1.5, 5))

# Soustraction des histogrammes
factor = len(flattened_opposite)/len(flattened_same)
hist_difference = hist_opposite - factor * hist_same

# Tracer l'histogramme
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
plt.step(bin_centers, hist_difference, where="mid", label="Opposé - Même charge", color="blue")
plt.xlabel('Masse invariante (GeV/c^2)')
plt.ylabel('Nombre d\'événements (Différence)')
plt.title('Différence des distributions des masses invariantes (0-5 GeV/c^2)')
plt.grid(True)
plt.legend()
plt.show()
```
